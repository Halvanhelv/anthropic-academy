```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from anthropic import Anthropic
```

```python
client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
```

```python
print(message.content[0].text)
```
